```python
def mean(array):
    n=len(array)
    return np.sum(array)/n
```

```python
mean(np.array([1,2,3,4,5,6]))
```

```python
def varience(array):
    n = len(array)
    m = mean(array)
    return np.sum((array - m)**2)/n
```

```python
varience(np.array([1,2,3,4,5,6]))
```

```python
def std(array):
    v = varience(array)
    return np.sqrt(v)
```

```python
std(np.array([1,2,3,4,5,6]))
```
